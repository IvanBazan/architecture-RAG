```python
#загружаем хранилище
searcher = ChromaSearch(PERSIST_DIRECTORY)
collection_info = searcher.get_collection_info(COLLECTION_NAME)

print(f"  Информация о коллекции '{COLLECTION_NAME}':")
print(f"  Количество чанков: {collection_info['count']}")
print(f"  Метрика расстояния: {collection_info['metadata']}")
```
