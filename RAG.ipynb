```python
#загружаем класс клиента Phi3
#проверка простых вопросов к LLM
import sys
sys.path.append('./src') 
from simple_phi3 import Phi3Client

client_simple = Phi3Client()
answer_simple = client_simple.ask("Столица России?")
print(answer_simple)
```

```python
#загружаем класс клиента Phi3 с RAG
#проверка простого вопроса на тему альпинизма из базы знаний
import sys
sys.path.append('./src') 
from RAG import Phi3RAGClient

client_rag = Phi3RAGClient()
question = "Где находится Пик Победы и какая у него высота?"
result = client_rag.ask(question)

print(result["answer"])

print("Источники:")
for source in result["sources"]:
    print(f"\t{source['source']}")
```

```python
#Сравнение ответов "чистой" Llm Phi-3 и RAG
#Вопросы на тему альпинизма

question = "Кто такие братья Троновы?"

print(question)
print("Ответ llm без использования RAG:")
answer_simple = client_simple.ask(question)
print(answer_simple)

print("Ответ RAG:")
answer_rag = client_rag.ask(question)
print(answer_rag["answer"])
```

```python
#загружаем класс клиента Phi3 с RAG с few_shot
#проверка простого вопроса на тему альпинизма из базы знаний
import sys
sys.path.append('./src') 
from RAG_fewShot import Phi3RAGClient_fs

client_rag_fs = Phi3RAGClient_fs()
question = "Кто такой Петр Семёнов Тянь-Шаньский?"
result = client_rag_fs.ask(question)

print(result["answer"])
```

```python
#загружаем класс клиента Phi3 с RAG с few_shot и CoT
#проверка простых вопросов на тему альпинизма из базы знаний

import sys
sys.path.append('./src') 
from RAG_CoT import Phi3RAGClient_CoT

client_rag_cot = Phi3RAGClient_CoT()

questions = [
    "Кто такой Петр Семёнов Тянь-Шанский?",
    "В каком году братья Троновы взошли на Белуху?",
    "Кто такие братья Абалаковы?"
]

for i, question in enumerate(questions, 1):
    print(f"Вопрос {i}: {question}")
    
    result = client_rag_cot.ask(question)
    print(f"Ответ:\n{result['answer']}")
    print()
```
